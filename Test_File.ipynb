```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression

# 1. إنشاء بيانات تجريبية بسيطة (مثلاً: المساحة بالمتر المربع وسعر الشقة)
# المساحة (متر مربع)
X = np.array([[60], [80], [100], [120], [140]])
# السعر (بالآلاف مثلاً)
y = np.array([300, 400, 500, 600, 700])

# 2. بناء نموذج الانحدار الخطي وتدريبه
model = LinearRegression()
model.fit(X, y)

# 3. التنبؤ بسعر شقة مساحتها 110 متر مربع
new_area = np.array([[110]])
predicted_price = model.predict(new_area)

print(f"المساحة المتوقعة للشقة (110 م²): {predicted_price[0]:.2f} ألف")
```
